```python
import pandas as pd 
import numpy as np 
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Flatten
from keras.layers import Conv1D
from keras.layers import MaxPooling1D
from tensorflow.keras.optimizers import Adam
```

```python
data = pd.read_csv("data_should_light.csv")
data = data.iloc[:,1:] 
y = data.iloc[:,-1] 
data = data.iloc[:,0:-1] 
```

```python
import sklearn 
from sklearn.preprocessing import StandardScaler 
from sklearn.model_selection import train_test_split 
```

```python
X,X_test,y,y_test = train_test_split(data.to_numpy(),y.to_numpy(),shuffle=True,train_size=0.8)
```

```python
print(X_test)
```

```python
model = Sequential([
    Dense(64, activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])


model.compile(optimizer="adam",
              loss='BinaryCrossentropy',
              metrics=['accuracy','mse'])
```

```python
model.fit(X, y, epochs=1000, verbose=1) 
```

```python
a = model.predict(X_test)
```

```python
import tensorflow as tf
a = tf.convert_to_tensor(a, dtype=tf.float32)
y_test = tf.convert_to_tensor(y_test, dtype=tf.float32)

# If `a` are probabilities, threshold them at 0.2
preds = tf.cast(a > 0.2, tf.float32)

# Accuracy
accuracy = tf.keras.metrics.BinaryAccuracy()
accuracy.update_state(y_test, preds)
acc_result = accuracy.result().numpy()

# MSE
mse = tf.keras.metrics.MeanSquaredError()
mse.update_state(y_test, a)
mse_result = mse.result().numpy()

print("Accuracy:", acc_result)
print("MSE:", mse_result)
```
